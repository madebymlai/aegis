<h2>GENERATING ALLOCATIONS USING ALL TICKERS IN DATA</h2>
THIS IS BASCILY OVERFITTING, AS WE ALLOW BUY TICKERS THAT WAS NOT IN THE NASDAQ 100 AT REBALANCING

```python
# Import necessary libraries
import numpy as np
import pandas as pd

# Get the close prices data
close_prices = data.get("Close")

# Define the allocation function to calculate performance between consecutive points
def top_performers_allocate_func(current_idx, previous_idx):
    # Calculate the percentage change between current and previous idx
    if previous_idx is not None:
        # Percentage change for each ticker
        pct_change = (close_prices.iloc[current_idx] / close_prices.iloc[previous_idx] - 1).dropna()

        # Rank tickers based on their percentage change and select the top 10
        top_10_tickers = pct_change.nlargest(10).index
        
        # Print top performers, their percentage change, and the date range used for calculation
        print(f"Top 10 Performers on {close_prices.index[current_idx].date()} from {close_prices.index[previous_idx].date()} to {close_prices.index[current_idx].date()}:")
        for ticker in top_10_tickers:
            print(f"{ticker}: {pct_change[ticker]*100:.2f}%")
    else:
        # If there is no previous index (first rebalancing), give no allocations
        top_10_tickers = []

    # Create an array of zeros for all tickers
    weights = np.zeros(symbol_wrapper.shape[1])
    
    # Assign 0.1 weight to the top 10 performing tickers
    for ticker in top_10_tickers:
        ticker_index = symbol_wrapper.columns.get_loc(ticker)
        weights[ticker_index] = 0.1

    return weights

# Apply allocations at each rebalancing point
previous_idx = None
for idx in ms_points:
    filled_allocations.iloc[idx] = top_performers_allocate_func(idx, previous_idx)
    previous_idx = idx  # Update previous_idx to the current idx

# Remove rows with NaN values
allocations_from_all_data = filled_allocations[~filled_allocations.isnull().any(axis=1)]

# Display final allocations
#allocations_from_all_data
```

<h2>GENERATING ALLOCATIONS USING THE NASDAQ 100 LIST</h2>
ONLY ALLOW TRADES TICKERS THAT WAS IN THE NASDAQ 100 LIST AT THE DATE WHEN REBALICING WAS DONE 

```python
import numpy as np
import pandas as pd
import pytz

# Load Nasdaq 100 index data
nasdaq100index = pd.read_csv('./data/nasdaq_100_index.csv')

# Convert Date column to datetime, localize to UTC and set time to midnight (00:00:00)
nasdaq100index['Date'] = pd.to_datetime(nasdaq100index['Date']).dt.tz_localize('UTC').dt.normalize()

# Get the close prices data
close_prices = data.get("Close")

# Define the allocation function to calculate performance between consecutive points
def top_performers_allocate_func(current_idx, previous_idx):
    # Get the current date from the close_prices index
    current_date = close_prices.index[current_idx].normalize()

    # Check if current_date exists in nasdaq100index['Date']
    matching_row = nasdaq100index[nasdaq100index['Date'] == current_date]
    if matching_row.empty:
        print(f"No tickers found for the date {current_date}.")
        return np.zeros(symbol_wrapper.shape[1])  # Return empty allocation if no tickers found

    # Get the tickers for the current date
    available_tickers = matching_row.iloc[:, 1:].values.flatten()
    available_tickers = available_tickers[~pd.isnull(available_tickers)]  # Remove NaN entries

    # Filter available_tickers to only those present in close_prices columns
    available_tickers_in_prices = [ticker for ticker in available_tickers if ticker in close_prices.columns]

    if len(available_tickers_in_prices) == 0:
        print(f"No available tickers found in close prices on {current_date}.")
        return np.zeros(symbol_wrapper.shape[1])  # Return empty allocation if no tickers available

    # Filter the close_prices dataframe for available tickers on the current date
    close_prices_filtered = close_prices[available_tickers_in_prices]

    # Calculate the percentage change between current and previous idx
    if previous_idx is not None:
        # Percentage change for each ticker
        pct_change = (close_prices_filtered.iloc[current_idx] / close_prices_filtered.iloc[previous_idx] - 1).dropna()

        # Rank tickers based on their percentage change and select the top 10
        top_10_tickers = pct_change.nlargest(10).index
        
        if top_10_tickers.empty:
            print(f"No top performers found for the date {current_date}.")
            return np.zeros(symbol_wrapper.shape[1])  # Return empty allocation if no top tickers found

        # Print top performers, their percentage change, and the date range used for calculation
        print(f"Top 10 Performers on {close_prices.index[current_idx].date()} from {close_prices.index[previous_idx].date()} to {close_prices.index[current_idx].date()}:")
        for ticker in top_10_tickers:
            print(f"{ticker}: {pct_change[ticker]*100:.2f}%")
    else:
        # If there is no previous index (first rebalancing), give no allocations
        top_10_tickers = []

    # Create an array of zeros for all tickers
    weights = np.zeros(symbol_wrapper.shape[1])
    
    # Assign 0.1 weight to the top 10 performing tickers
    for ticker in top_10_tickers:
        ticker_index = symbol_wrapper.columns.get_loc(ticker)
        weights[ticker_index] = 0.1

    return weights

# Apply allocations at each rebalancing point
previous_idx = None
for idx in ms_points:
    filled_allocations.iloc[idx] = top_performers_allocate_func(idx, previous_idx)
    previous_idx = idx  # Update previous_idx to the current idx

# Remove rows with NaN values
allocations_from_only_nasdaq100_index = filled_allocations[~filled_allocations.isnull().any(axis=1)]

# Display final allocations
#print(allocations_from_only_nasdaq100_index)
```
