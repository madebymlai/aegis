# Indicator Playbook Example
One indicator idea/family with parameter sweeps.

```python
windows = [10, 20, 50]
wtypes = ['simple', 'wilder']
thresholds = [0.0, 0.01]
variant_records = []
for window in windows:
    for wtype in wtypes:
        for threshold in thresholds:
            variant_records.append({
                'variant_id': f'ma-{window}-{wtype}-{threshold}',
                'indicator_source': 'playbook',
                'indicator_id': 'example_ma_explore',
                'params': {'window': window, 'wtype': wtype, 'threshold': threshold},
                'metrics': {'total_return_pct': 0.0},
                'baseline_component_indicator_id': 'example.ma',
            })
AEGIS_PLAYBOOK_RESULT = {'variant_records': variant_records}
```
